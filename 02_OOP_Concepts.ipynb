#  – Object-Oriented Programming

## 1. Class

A class is a blueprint for creating objects.

```python
class Student:
    name = "Sai"
    age = 25
```

## 2. Object

An object is an instance of a class.

```python
class Student:
    name = "Sai"

student1 = Student()

print(student1.name)
```

## 3. Constructor `__init__()`

The `__init__()` method runs automatically when an object is created.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student = Student("Sai", 25)

print(student.name)
print(student.age)
```

## 4. Instance Variables

Instance variables belong to a specific object.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("Sai", 25)
s2 = Student("Ravi", 24)

print(s1.name)
print(s2.name)
```

## 5. Class Variables

Class variables are shared by all objects.

```python
class Student:
    college = "ABC College"

    def __init__(self, name):
        self.name = name

s1 = Student("Sai")
s2 = Student("Ravi")

print(s1.college)
print(s2.college)
```

## 6. Instance Methods

Instance methods work with object data.

```python
class Student:
    def __init__(self, name):
        self.name = name

    def display(self):
        print("Student:", self.name)

student = Student("Sai")
student.display()
```

## 7. Static Methods

A static method does not need `self` or `cls`.

```python
class Calculator:

    @staticmethod
    def add(a, b):
        return a + b

print(Calculator.add(10, 20))
```

## 8. Class Methods

A class method works with the class itself and receives `cls`.

```python
class Student:
    college = "ABC College"
    @classmethod
    def change_college(cls, name):
        cls.college = name
Student.change_college("XYZ College")
print(Student.college)
```

## 9. Properties `@property`

Properties allow a method to be accessed like an attribute.

```python
class Student:
    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name

student = Student("Sai")

print(student.name)
```

## 10. Encapsulation

Encapsulation means keeping data and methods together and controlling access to data.

```python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

account = BankAccount(5000)

print(account.get_balance())
```

## 11. Abstraction

Abstraction hides implementation details and exposes only the necessary functionality.

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass


class Car(Vehicle):

    def start(self):
        print("Car started")

car = Car()
car.start()
```

## 12. Inheritance

Inheritance allows one class to reuse another class's functionality.

```python
class Animal:
    def eat(self):
        print("Eating")


class Dog(Animal):
    def bark(self):
        print("Barking")


dog = Dog()

dog.eat()
dog.bark()
```

## 13. Multiple Inheritance

Multiple inheritance allows a class to inherit from more than one parent class.

```python
class Father:
    def skills(self):
        print("Business")


class Mother:
    def talent(self):
        print("Music")


class Child(Father, Mother):
    pass


child = Child()

child.skills()
child.talent()
```

## 14. Multilevel Inheritance

Multilevel inheritance occurs across multiple inheritance levels.

```python
class Grandparent:
    def house(self):
        print("Grandparent's house")


class Parent(Grandparent):
    def car(self):
        print("Parent's car")


class Child(Parent):
    def bike(self):
        print("Child's bike")


child = Child()

child.house()
child.car()
child.bike()
```

## 15. Hierarchical Inheritance

Hierarchical inheritance occurs when multiple classes inherit from the same parent class.

```python
class Animal:
    def eat(self):
        print("Eating")
class Dog(Animal):
    def bark(self):
        print("Barking")
class Cat(Animal):
    def meow(self):
        print("Meowing")

dog = Dog()
cat = Cat()
dog.eat()
cat.eat()
```

## 16. Polymorphism

Polymorphism means the same interface can produce different behavior.

```python
class Dog:
    def sound(self):
        print("Bark")
class Cat:
    def sound(self):
        print("Meow")
animals = [Dog(), Cat()]
for animal in animals:
    animal.sound()
```

## 17. Method Overriding

Method overriding occurs when a child class provides its own version of a parent method.

```python
class Animal:
    def sound(self):
        print("Animal sound")
class Dog(Animal):
    def sound(self):
        print("Bark")
dog = Dog()
dog.sound()
```

## 18. Method Overloading – Python Approach

Python does not support traditional method overloading by defining multiple methods with the same name. Default arguments or `*args` are commonly used.

```python
class Calculator:

    def add(self, a, b=0, c=0):
        return a + b + c
calc = Calculator()
print(calc.add(10))
print(calc.add(10, 20))
print(calc.add(10, 20, 30))
```

### Method Overloading with `*args`

Using `*args`, a method can accept a variable number of arguments.

```python
class Calculator:
    def add(self, *numbers):
        return sum(numbers)
calc = Calculator()
print(calc.add(10, 20))
print(calc.add(10, 20, 30, 40))
```

## 19. Composition

Composition represents a strong **has-a** relationship. The containing object creates and owns another object.

```python
class Engine:
    def start(self):
        print("Engine started")
class Car:
    def __init__(self):
        self.engine = Engine()
    def start(self):
        self.engine.start()
car = Car()
car.start()
```

## 20. Aggregation

Aggregation represents a weaker **has-a** relationship where the related object can exist independently.

```python
class Teacher:
    def __init__(self, name):
        self.name = name
class Department:
    def __init__(self, teacher):
        self.teacher = teacher
teacher = Teacher("Ravi")
department = Department(teacher)
print(department.teacher.name)
```

## 21. Association

Association represents a general relationship between independent objects.

```python
class Student:
    def study(self):
        print("Student is studying")
class Teacher:
    def teach(self):
        print("Teacher is teaching")
student = Student()
teacher = Teacher()
teacher.teach()
student.study()
```

## 22. Magic / Dunder Methods

Magic (dunder) methods are special methods surrounded by double underscores. Examples include `__init__`, `__str__`, `__repr__`, `__len__`, `__add__`, and `__eq__`.

```python
class Demo:
    def __init__(self, value):
        self.value = value
    def __str__(self):
        return str(self.value)
demo = Demo(10)
print(demo)
```

## 23. `super()`

`super()` is used to call functionality from a parent class.

```python
class Parent:
    def __init__(self):
        print("Parent constructor")
class Child(Parent):
    def __init__(self):
        super().__init__()
        print("Child constructor")
child = Child()
```

## 24. `__str__()`

`__str__()` defines the user-friendly string representation of an object.

```python
class Student:
    def __init__(self, name):
        self.name = name
    def __str__(self):
        return f"Student: {self.name}"
student = Student("Sai")
print(student)
```

## 25. `__repr__()`

`__repr__()` provides a developer-oriented representation of an object.

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f"Student(name='{self.name}', age={self.age})"


student = Student("Sai", 25)

print(repr(student))
```

## 26. `__len__()`

`__len__()` defines what `len(object)` should return.

```python
class Team:

    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)


team = Team(["Sai", "Ravi", "Kumar"])

print(len(team))
```

## 27. Dataclasses — Bonus

The `@dataclass` decorator automatically generates common methods such as `__init__()` and `__repr__()`.

```python
from dataclasses import dataclass
@dataclass
class Student:
    name: str
    age: int
    course: str
student = Student("Sai", 25, "Data Science")
print(student)
```

## 28. Singleton Pattern — Bonus

The Singleton pattern ensures that only one instance of a class is created.

```python
class Singleton:
    _instance = None
    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance
obj1 = Singleton()
obj2 = Singleton()
print(obj1 is obj2)
```
