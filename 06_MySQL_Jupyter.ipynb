## 5. Tables
Create employees and departments tables.

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS departments (
    id INT PRIMARY KEY,
    department_name VARCHAR(50)
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS employees (
    id INT PRIMARY KEY,
    name VARCHAR(50),
    city VARCHAR(50),
    salary INT,
    dept_id INT,
    FOREIGN KEY (dept_id) REFERENCES departments(id)
)
""")

conn.commit()
print("Tables created!")
```

## 6. Primary Key
A primary key uniquely identifies each row.

```python
cursor.execute("SHOW CREATE TABLE employees")
print(cursor.fetchone()[1])
```

## 7. Foreign Key
A foreign key connects employees to departments.

```python
cursor.execute("""
SELECT e.name, d.department_name
FROM employees e
JOIN departments d ON e.dept_id = d.id
""")

for row in cursor.fetchall():
    print(row)
```

## 8. Relationships
Employees and departments are connected through `dept_id`.

```python
cursor.execute("""
SELECT e.name, d.department_name
FROM employees e
INNER JOIN departments d
ON e.dept_id = d.id
""")

for row in cursor.fetchall():
    print(row)
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS test_constraints (
    id INT PRIMARY KEY,
    name VARCHAR(50) NOT NULL,
    age INT CHECK (age >= 18)
)
""")
conn.commit()
print("Constraints created!")
```

## 11. SELECT
Retrieve data from a table.

```python
cursor.execute("SELECT * FROM employees")
for row in cursor.fetchall():
    print(row)
```

## 12. WHERE
Filter rows using a condition.

```python
cursor.execute(
    "SELECT * FROM employees WHERE salary > %s",
    (40000,)
)
for row in cursor.fetchall():
    print(row)
```

## 13. ORDER BY
Sort the result by salary.

```python
cursor.execute(
    "SELECT name, salary FROM employees ORDER BY salary DESC"
)
for row in cursor.fetchall():
    print(row)
```

## 14. GROUP BY
Group rows for calculations.

```python
cursor.execute("""
SELECT dept_id, COUNT(*) AS total
FROM employees
GROUP BY dept_id
""")
for row in cursor.fetchall():
    print(row)
```

## 15. HAVING
Filter grouped results.

```python
cursor.execute("""
SELECT dept_id, AVG(salary) AS avg_salary
FROM employees
GROUP BY dept_id
HAVING AVG(salary) > 40000
""")
for row in cursor.fetchall():
    print(row)
```

## 16. DISTINCT
Return unique cities.

```python
cursor.execute("SELECT DISTINCT city FROM employees")
for row in cursor.fetchall():
    print(row)
```

## 17. LIMIT
Return only a few rows.

```python
cursor.execute("SELECT * FROM employees LIMIT 3")
for row in cursor.fetchall():
    print(row)
```

## 18. INNER JOIN
Return matching rows from both tables.

```python
cursor.execute("""
SELECT e.name, d.department_name
FROM employees e
INNER JOIN departments d ON e.dept_id = d.id
""")
for row in cursor.fetchall():
    print(row)
```

## 19. LEFT JOIN
Return all employees and matching departments.

```python
cursor.execute("""
SELECT e.name, d.department_name
FROM employees e
LEFT JOIN departments d ON e.dept_id = d.id
""")
for row in cursor.fetchall():
    print(row)
```

## 20. RIGHT JOIN
Return all departments and matching employees.

```python
cursor.execute("""
SELECT e.name, d.department_name
FROM employees e
RIGHT JOIN departments d ON e.dept_id = d.id
""")
for row in cursor.fetchall():
    print(row)
```

## 21. FULL JOIN
MySQL does not directly support FULL OUTER JOIN. Use `UNION` of LEFT and RIGHT JOIN.

```python
cursor.execute("""
SELECT e.name, d.department_name
FROM employees e
LEFT JOIN departments d ON e.dept_id = d.id

UNION

SELECT e.name, d.department_name
FROM employees e
RIGHT JOIN departments d ON e.dept_id = d.id
""")
for row in cursor.fetchall():
    print(row)
```

## 22. CROSS JOIN
Return every employee-department combination.

```python
cursor.execute("""
SELECT e.name, d.department_name
FROM employees e
CROSS JOIN departments d
LIMIT 10
""")
for row in cursor.fetchall():
    print(row)
```

## 23. UNION
Combine two compatible result sets.

```python
cursor.execute("""
SELECT city FROM employees WHERE city = 'Hyderabad'
UNION
SELECT city FROM employees WHERE city = 'Chennai'
""")
for row in cursor.fetchall():
    print(row)
```

## 24. CASE
Create values based on conditions.

```python
cursor.execute("""
SELECT name, salary,
CASE
    WHEN salary >= 50000 THEN 'High'
    WHEN salary >= 40000 THEN 'Medium'
    ELSE 'Low'
END AS salary_level
FROM employees
""")
for row in cursor.fetchall():
    print(row)
```

## 25. Subquery
Use one query inside another query.

```python
cursor.execute("""
SELECT name, salary
FROM employees
WHERE salary > (SELECT AVG(salary) FROM employees)
""")
for row in cursor.fetchall():
    print(row)
```

## 26. CTE
A CTE gives a temporary name to a query result.

```python
cursor.execute("""
WITH high_salary AS (
    SELECT * FROM employees
    WHERE salary > 40000
)
SELECT * FROM high_salary
""")
for row in cursor.fetchall():
    print(row)
```

## 27. View
A view stores a reusable SQL query.

```python
cursor.execute("DROP VIEW IF EXISTS high_salary")
cursor.execute("""
CREATE VIEW high_salary AS
SELECT name, salary
FROM employees
WHERE salary > 40000
""")
cursor.execute("SELECT * FROM high_salary")
for row in cursor.fetchall():
    print(row)
```

## 28. Index
Create an index to help searches on a column.

```python
cursor.execute("CREATE INDEX idx_employee_name ON employees(name)")
conn.commit()
print("Index created!")
```

## 29. ROW_NUMBER()
Give each employee a unique row number.

```python
cursor.execute("""
SELECT name, salary,
ROW_NUMBER() OVER (ORDER BY salary DESC) AS row_num
FROM employees
""")
for row in cursor.fetchall():
    print(row)
```

## 30. RANK()
Give equal salaries the same rank, with gaps.

```python
cursor.execute("""
SELECT name, salary,
RANK() OVER (ORDER BY salary DESC) AS salary_rank
FROM employees
""")
for row in cursor.fetchall():
    print(row)
```

## 31. DENSE_RANK()
Give equal salaries the same rank without gaps.

```python
cursor.execute("""
SELECT name, salary,
DENSE_RANK() OVER (ORDER BY salary DESC) AS salary_rank
FROM employees
""")
for row in cursor.fetchall():
    print(row)
```

## 32. LAG()
Get the salary from the previous row.

```python
cursor.execute("""
SELECT name, salary,
LAG(salary) OVER (ORDER BY salary) AS previous_salary
FROM employees
""")
for row in cursor.fetchall():
    print(row)
```

## 33. LEAD()
Get the salary from the next row.

```python
cursor.execute("""
SELECT name, salary,
LEAD(salary) OVER (ORDER BY salary) AS next_salary
FROM employees
""")
for row in cursor.fetchall():
    print(row)
```
