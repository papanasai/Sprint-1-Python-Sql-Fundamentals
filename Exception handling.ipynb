# Exception Handling

```python
# try and except
try:
    a = 10
    b = 0
    result = a / b
    print(result)

except ZeroDivisionError:
    print("Cannot divide by zero")
```

# else
# else runs only when there is no exception.

```python
try:
    a = 100
    b = 50
    result = a/b
    print(100/50)
except zeroDivisionError:
    print("cannot divide by zero")

else :
    print("calculation sucessfull")
```

# finally
# finally runs whether an exception occurs or not.


```python
try:
    print("program started")

except:
    print("error occured")

finally:
    print("program ended")
    
```

# Raise 
# raise is used to manually generate an exception
 

```python
age = 28
if age < 18:
    raise ValueError("Age must be 18 or above")
```

# 1. assert
# assert is used to check whether a condition is true.

```python
age = 35
assert age >= 18, "Age must be 18 or above"
print("Eligible")
```

# ZeroDivisionError
# Meaning: Trying to divide a number by zero.

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")
```

# ValueError
# Meaning: Correct data type, but invalid value

```python
try:
    number = int("abc")
except ValueError:
    print("Invalid value")
```

# TypeError
# Meaning: Operation is performed on an incompatible data type

```python
try:
    result = "10" + 5
except TypeError:
    print("Invalid data type")
```

# NameError
# Using a variable that has not been defined.

```python
try:
    print(name)
except NameError:
    print("Variable does not exist")
```

# IndexError
# Meaning: Trying to access a list index that doesn't exist


```python
numbers = [10, 20, 30]
try:
    print(numbers[5])
except IndexError:
    print("Invalid list index")
```

# KeyError
# Meaning: Trying to access a dictionary key that doesn't exist.

```python
student = {
    "name": "Sai",
    "age": 25
}
try:
    print(student["salary"])
except KeyError:
    print("Dictionary key does not exist")
```

# FileNotFoundError
# Meaning: Trying to open a file that doesn't exist.

```python
try:
    file = open("data.txt", "r")
except FileNotFoundError:
    print("File does not exist")
```

# AttributeError
# Meaning: Trying to use an attribute or method that doesn't exist.

```python
name = "Sai"
try:
    name.append("Kumar")
except AttributeError:
    print("Invalid attribute or method")
```
