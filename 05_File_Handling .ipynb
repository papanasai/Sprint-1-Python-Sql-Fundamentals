# File Handling
Short Python examples for Jupyter Notebook.

## 1. Read a Text File
Use `open()` to read a file.

```python
with open("sample.txt", "w") as f:
    f.write("Hello Python!")

with open("sample.txt", "r") as f:
    print(f.read())
```

## 2. Read Lines
`readline()` reads one line. `readlines()` reads all lines.

```python
with open("sample.txt", "r") as f:
    print(f.readline())
    print(f.readlines())
```

## 3. Write a File
`w` creates or replaces a file.

```python
with open("output.txt", "w") as f:
    f.write("First line\nSecond line")

print("File written")
```

## 4. Append to a File
`a` adds content at the end.

```python
with open("output.txt", "a") as f:
    f.write("\nThird line")

print("Content added")
```

## 5. CSV File
CSV stores tabular data.

```python
import csv

rows = [
    ["Name", "Age"],
    ["Sai", 25],
    ["Ravi", 24]
]

with open("students.csv", "w", newline="") as f:
    csv.writer(f).writerows(rows)

print("CSV created")
```

### Read CSV

```python
import csv

with open("students.csv") as f:
    for row in csv.reader(f):
        print(row)
```

## 6. JSON File
JSON is commonly used for APIs and data exchange.

```python
import json

student = {"name": "Sai", "age": 25}

with open("student.json", "w") as f:
    json.dump(student, f, indent=2)

print("JSON created")
```

### Read JSON

```python
import json

with open("student.json") as f:
    data = json.load(f)

print(data)
```

## 7. Pickle
Pickle saves Python objects in binary format.

```python
import pickle

data = {"name": "Sai", "skills": ["Python", "SQL"]}

with open("student.pkl", "wb") as f:
    pickle.dump(data, f)

print("Pickle created")
```

### Read Pickle

```python
import pickle

with open("student.pkl", "rb") as f:
    print(pickle.load(f))
```

## 8. pathlib
`Path` makes file and folder handling simple.

```python
from pathlib import Path

file = Path("sample.txt")
print(file.name)
print(file.suffix)
print(file.exists())
```

### Create and Read with pathlib

```python
from pathlib import Path

file = Path("example.txt")
file.write_text("Hello pathlib!")
print(file.read_text())
```

## 9. os Module
`os` can check directories and paths.

```python
import os

print(os.getcwd())
print(os.listdir("."))
```

## 10. shutil
`shutil` can copy and move files.

```python
import shutil

shutil.copy("sample.txt", "copy.txt")
print("File copied")
```

## 11. File Validation
Check whether a file exists and is not empty.

```python
from pathlib import Path

file = Path("sample.txt")

if file.exists() and file.is_file() and file.stat().st_size > 0:
    print("Valid file")
else:
    print("Invalid file")
```

## 12. Find Files
`glob()` finds files by pattern.

```python
from pathlib import Path

for file in Path(".").glob("*.txt"):
    print(file)
```

## 13. Exception Handling
Use `try-except` for safe file operations.

```python
try:
    with open("missing.txt") as f:
        print(f.read())
except FileNotFoundError:
    print("File not found")
```

## 14. Practical Example
Save student data as JSON and read it back.

```python
import json
from pathlib import Path

file = Path("student_data.json")
student = {"name": "Sai", "course": "Data Science"}

file.write_text(json.dumps(student, indent=2))
print(json.loads(file.read_text()))
```

## 15. Backup File
Create a simple backup using `shutil`.

```python
from pathlib import Path
import shutil

source = Path("student_data.json")
backup = Path("student_data_backup.json")

if source.exists():
    shutil.copy2(source, backup)
    print("Backup created")
```
