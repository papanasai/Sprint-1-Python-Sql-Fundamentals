# A while loop repeatedly executes a block of code as long as a condition True

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

# break loop 
# break immediately stops the loop.

```python
count = 1

while count <= 10:

    if count == 5:
        break

    print(count)
    count += 1
```

# continue Loop skips the current iteration.


```python
count = 0
while count < 5:
    count += 1
    if count == 3:
        continue
    print(count)
```

# Arguments
 # a arguments are the values or objects passed into a function when it is called parameter

```python
# Default Arguments
def greet(name, message="Hello"):
    print(f"{message}, {name}!")
greet("Bob")             
greet("Bob", "Goodbye")
```

```python
# Lamda functions 
greet = lambda: "Hello, World!"
print(greet())
```

```python
# Recursive Functions 
def factorial(n):
    if n == 1:
        return 1
    else:
        return n * factorial(n - 1)
print(factorial(4))
```

# Yeild 
# Creating and Using a Generator is called yeild 

```python
# yield is used inside a function to create a generator
def numbers():
    yield 1
    yield 2
    yield 3

for num in numbers():
    print(num)
```

# Decorators
# A decorator is a function that adds extra functionality to another function without changing its original code.

```python
def decorator(func):
    def wrapper():
        print("Before function")
        func()
        print("After function")    
    return wrapper
@decorator
def greet():
    print("Hello!")
greet()
```

```python
# closures
# Closure = A function that remembers variables from its outer function.
def outer():
    message = "Hello"
    def inner():
        print(message)
    return inner
my_function = outer()
my_function()
```

```python
# global and nonlocal
# a variable defined outside the function at the global level.
def outer():
    x = 10
    def inner():
        nonlocal x
        x = 20
    inner()
    print(x)
outer()
```
