# Logging



## 1. What is Logging?
# Logging is the process of recording events that happen during the execution of a program.
# It helps developers track the flow of execution, errors, warnings, and other runtime information.

# Why Use Logging

Provides visibility into what your program is doing.
Helps in debugging and troubleshooting issues.
Useful for monitoring applications in production.
Allows configurable levels of detail (verbosity).

# 3. `print()` vs `logging

Aspect	            print()	                               logging
Purpose  	   Displays output to console          	Records structured logs with levels
Flexibility	       Limited	                          Highly configurable (levels, handlers, formatters)
Control	       No control over severity	                 Control via log levels
Output           Always prints   	                 Can be directed to files, console, or external systems

# 1.DEBUG Logging

```python
import logging

# Configure logging to show DEBUG and above
logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

# Example debug usage
logging.debug("Starting the program...")

x = 10
y = 5
logging.debug(f"Values before calculation: x={x}, y={y}")

result = x / y
logging.debug(f"Calculation result: {result}")
```

# 2 Info Logging

```python
import logging
logging.basicConfig(level=logging.INFO)
logging.info('Application started')
```

# 3.Warning Logging

```python
logging.warning('Low disk space')
```

```python
# ERROR Logging
import logging

# Configure logging
logging.basicConfig(
    level=logging.ERROR,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

try:
    result = 10 / 0
except ZeroDivisionError:
    logging.error("Division by Zero occurred")
```

# 5 CRITICAL logging

```python
logging.critical("Critical failure occurred!")
```

```python
# Log Formaters
formatter = logging.Formatter("%(asctime)s - %(name)s - %(levelname)s - %(filename)s:%(lineno)d - %(message)s")
```

# File handlers

```python
file_handler = logging.FileHandler("application.log")   
file_handler.setLevel(logging.ERROR)                    
file_handler.setFormatter(formatter) 
```

# Console Handlers 

```python
import logging

logger = logging.getLogger(__name__)

console_handler = logging.StreamHandler()
console_handler.setLevel(logging.INFO)

logger.addHandler(console_handler)

logger.info("Application started")
logger.error("Something went wrong")
```

# Rotating Logs

```python
import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger("app")
logger.setLevel(logging.INFO)

logger.addHandler(RotatingFileHandler("app.log", maxBytes=1000, backupCount=3))
logger.addHandler(logging.StreamHandler())

logger.info("App started")
logger.error("Something went wrong")
```
